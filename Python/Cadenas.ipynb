# Cadenas en Python
Las cadenas son secuencias de caracteres y se pueden definir usando comillas simples (' '), comillas dobles (" ") o triple comillas (""" """ o ''' ''').

## “””text”””
Se usa triple doble comillas para texto de varias lineas.

## Algunas Funciones de Cadenas
- `lower()`: Aplica lowercase a la var_string
- `upper()`: Aplica uppercase a la var_string
- `capitalize()`: Aplica capitalize a la var_string
- `title()`: Aplica title a la var_string
- `strip()`: Elimina espacios al inicio y al final de la var_string
- `replace(old, new)`: Reemplaza el texto old por new en la var_string
- `split(sep)`: Divide la var_string en una lista usando el separador sep
- `splitlines()`: Divide la var_string en una lista usando el salto de línea como separador (\n)
- `join(list)`: Une los elementos de la lista en una cadena, usando el separador que se aplica al método. Por ejemplo: `", ".join(list)` une los elementos de la lista con una coma y un espacio.
- `find(substring, start, end)`: Retorna el índice de la primera aparición del substring en la var_string, si no se encuentra retorna -1. Los parámetros start y end son opcionales y permiten buscar en un rango específico de la cadena.
- `index(substring, start, end)`: Similar a find(), pero lanza un error si el substring no se encuentra.
- `replace(old, new, count)`: Reemplaza el texto old por new en la var_string, pero solo reemplaza count veces. Si count no se especifica, reemplaza todas las ocurrencias.

## Formateo de Cadenas
- `f"texto {variable}"`: Permite insertar variables dentro de una cadena de texto.
- `f"texto {variable:format}"`: Permite insertar variables dentro de una cadena de texto y aplicar un formato específico. Por ejemplo, `f"El número es {num:.2f}"` formatea el número a dos decimales.
- `"texto {}".format(variable)`: Similar al anterior, pero usa el método format() para insertar variables.
- `"texto {name_1} {name_2}".format(name_1=variable_1, name_2=variable_2)`: Permite insertar múltiples variables en una cadena de texto usando el método format().

## Templates
- `Template`: Permite crear plantillas de texto con variables que se pueden reemplazar. Se importa desde el módulo `string`.
- `Template("texto $variable").substitute(variable=valor)`: Reemplaza la variable $variable en el texto por el valor proporcionado. Si la variable no se encuentra, lanza un error.#%% md


```python
from string import Template

template = Template("Hola, $name. Tu edad es $age.")
result = template.substitute(name="Alice", age=30)

print(result)  # Output: Hola, Alice. Tu edad es 30.
```

## Expresiones Regulares
Las expresiones regulares (regex) son patrones que se utilizan para buscar y manipular texto. En Python, se utilizan a través del módulo `re`.

Ejemplo: `r'st\d\s\w{3,10}'` la cadena debe comenzar con "st", seguido de un dígito, un espacio, y luego una palabra de 3 a 10 caracteres alfanuméricos.

Metacaracteres:
- `\d`: Representa un dígito (0-9).
- `\D`: Representa cualquier carácter que no sea un dígito.
- `\w`: Representa un carácter alfanumérico (letras, dígitos y guiones bajos).
- `\W`: Representa cualquier carácter que no sea alfanumérico.
- `\s`: Representa un espacio en blanco (espacio, tabulación, salto de línea).
- `\S`: Representa cualquier carácter que no sea un espacio en blanco.
- `.`: Representa cualquier carácter excepto un salto de línea.
- `*`: Indica que el elemento anterior puede aparecer cero o más veces.
- `+`: Indica que el elemento anterior debe aparecer al menos una vez.
- `?`: Indica que el elemento anterior es opcional (puede aparecer cero o una vez).
- `{n}`: Indica que el elemento anterior debe aparecer exactamente n veces.
- `$`: Indica el final de la cadena. Es decir, el patrón debe coincidir con el final de la cadena.
- `\`: Escapa un carácter especial para que sea tratado como un carácter literal. Por ejemplo, `\.` busca un punto literal en lugar de cualquier carácter.
- `|`: Indica una alternativa, es decir, el patrón puede coincidir con uno u otro. Por ejemplo, `a|b` coincide con "a" o "b".
- `[]`: Define un conjunto de caracteres. Por ejemplo, `[abc]` coincide con "a", "b" o "c". También se pueden usar rangos como `[a-z]` para letras minúsculas o `[0-9]` para dígitos.
- `^`: Indica el inicio de la cadena. Es decir, el patrón debe coincidir con el inicio de la cadena.

### Funciones Comunes de `re`
- `re.search(pattern, string)`: Busca el patrón en la cadena y retorna un objeto
- `re.match(pattern, string)`: Busca el patrón al inicio de la cadena y retorna un objeto
- `re.fullmatch(pattern, string)`: Busca el patrón en toda la cadena y retorna un objeto
- `re.findall(pattern, string)`: Retorna una lista con todas las coincidencias del patrón en la cadena
- `re.sub(pattern, replacement, string)`: Reemplaza todas las coincidencias del patrón en la cadena por el texto de reemplazo
- `re.split(pattern, string)`: Divide la cadena en una lista usando el patrón como separador
- `re.compile(pattern)`: Compila el patrón en un objeto regex para su reutilización.
- `re.sub(pattern, replacement, string)`: Reemplaza todas las coincidencias del patrón en la cadena por el texto de reemplazo.

```python
import re

pattern = r'st\d\s\w{3,10}'

text = "st1 abcdefghij st2 klmno st3 pqrstuvwxyz"

# Buscar coincidencias
matches = re.findall(pattern, text)
print(matches)  # Output: ['st1 abcdefghij', 'st2 klmno', 'st3 pqrstuvwxyz']

# Reemplazar coincidencias
replacement = "REPLACED"
new_text = re.sub(pattern, replacement, text)
print(new_text)  # Output: "REPLACED REPLACED REPLACED"

# Dividir la cadena
split_text = re.split(pattern, text)
print(split_text)  # Output: ['', ' REPLACED ', ' REPLACED ', ' REPLACED']

# Repeticion de patrones
repeated_pattern1 = r'\d{5}' # Busca 5 dígitos consecutivos.
repeated_pattern4 = r'\d{1,4}' # Busca de 1 a 4 dígitos consecutivos.
repeated_pattern2 = r'\w+' # Busca una o más letras, números o guiones bajos.
repeated_pattern3 = r'\s?' # Busca un espacio en blanco opcional, es decir, puede haber UN espacio o no.


```
