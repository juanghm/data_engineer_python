# Fechas

## Clase date
Se pueden crear fechas con la clase `date` del módulo `datetime`. Esta clase permite trabajar con fechas de manera sencilla y eficiente. Lo que permite obtener y manipular fechas de forma fácil.

```python
from datetime import date

# Crear una fecha específica
fecha_especifica = date(2023, 10, 1)  # Año, Mes, Día

# Imprimir la fecha
print(fecha_especifica)  # Output: 2023-10-01

# Obtener el dia de la semana.
print(fecha_especifica.weekday())  # Output: 6 (Domingo es 6, Lunes es 0, etc.)

# Obtener el mes y el año
print(fecha_especifica.month)  # Output: 10
print(fecha_especifica.year)  # Output: 2023
```

### Calcular la diferencia entre dos fechas
Para calcular la diferencia entre dos fechas, se pueden restar directamente dos objetos `date`. El resultado es un objeto `timedelta` que representa la diferencia en días.

```python
from datetime import timedelta

# Crear dos fechas
fecha_inicio = date(2023, 10, 1)
fecha_fin = date(2023, 10, 15)

# Calcular la diferencia
diferencia = fecha_fin - fecha_inicio
print(diferencia)  # Output: 14 days, 0:00:00

# Obtener la cantidad de días
print(diferencia.days)  # Output: 14

# Crear un delta de tiempo de 10 días
delta_tiempo = timedelta(days=10)

# Sumar el delta de tiempo a la fecha de inicio
nueva_fecha = fecha_inicio + delta_tiempo
print(nueva_fecha)  # Output: 2023-10-11
```

### Convirtiendo fechas a cadenas de texto
Para convertir un objeto `date` a una cadena de texto, se puede usar el método `strftime()`, que permite formatear la fecha según un formato específico.

```python
from datetime import date

# Crear una fecha
fecha = date(2023, 10, 1)

print(fecha) # Output: 2023-10-01

print(fecha.isoformat()) # Output: 2023-10-01

print(fecha.strftime("%Y")) # Output: 2023

print(fecha.strftime("%Y/%m")) # Output: 2023/10
```

## Fechas con hora, minuto, segundo y microsegundo (datetime)
La clase `datetime` del módulo `datetime` permite trabajar con fechas y horas. Esta clase incluye información sobre el año, mes, día, hora, minuto, segundo y microsegundo.

```python
from datetime import datetime
fecha = datetime(2023, 10, 1, 15, 23, 25, 500000)  # Año, Mes, Día, Hora, Minuto, Segundo, Microsegundo
print(fecha)

new_fecha = fecha.replace(year=2024, minute=12)  # Cambiar el año a 2024
print(new_fecha)
```

### Convirtiendo cadenas de texto a fechas
Para convertir una cadena de texto a un objeto `datetime`, se puede usar el método `strptime()`, que permite especificar el formato de la cadena.

```python
from datetime import datetime

# Convertir una cadena de texto a un objeto datetime
fecha_str = "2023-10-01 15:23:25"
fecha_datetime = datetime.strptime(fecha_str, "%Y-%m-%d %H:%M:%S")

# Timestamp a fecha
timestamp = 1696152205.123456
fecha_datetime_from_timestamp = datetime.fromtimestamp(timestamp)
print(fecha_datetime)  # Output: 2023-10-01 15:23:25
```

### Duracion entre dos fechas y horas
Para calcular la duración entre dos objetos `datetime`, se pueden restar directamente. El resultado es un objeto `timedelta` que representa la diferencia en días, segundos y microsegundos.

```python
from datetime import datetime

# Crear dos fechas y horas
fecha_inicio = datetime(2023, 10, 1, 15, 23, 25)
fecha_fin = datetime(2023, 10, 15, 18, 30, 0)

# Calcular la diferencia
diferencia = fecha_fin - fecha_inicio
print(diferencia)  # Output: 14 days, 3:06:35
# Obtener la cantidad de días, segundos y microsegundos
print(diferencia.days)  # Output: 14
print(diferencia.seconds)  # Output: 11195 (3 horas, 6 minutos y 35 segundos)
```

### UTC y zonas horarias
Para trabajar con zonas horarias, se puede usar el módulo `pytz` junto con `datetime`. Esto permite manejar fechas y horas en diferentes zonas horarias de manera sencilla.

```python
from datetime import datetime, timedelta, timezone

et = timezone(timedelta(hours=-5)) # Hora estándar del Este (EST)
fecha_et = datetime(2023, 10, 1, 15, 23, 25, tzinfo=et)  # Hora en EST

print(fecha_et)

ist = timezone(timedelta(hours=5, minutes=30))  # Hora estándar de la India (IST)
print(fecha_et.astimezone(ist))  # Convertir a IST

print(fecha_et.replace(tzinfo=timezone.utc)) # Convertir a UTC

print(fecha_et.astimezone(timezone.utc)) # Convertir a UTC usando astimezone
```

### dateutil tz
Para manejar zonas horarias de manera más avanzada, se puede usar el módulo `dateutil.tz`. Este módulo permite trabajar con zonas horarias de manera más flexible y sencilla.

- tz.datetime_ambiguous(): Permite manejar ambigüedades en las zonas horarias, como el horario de verano.
- tz.enfold(): Permite crear un objeto `datetime` con una zona horaria específica, evitando ambigüedades.

```python
from dateutil import tz

# Crear una zona horaria para Nueva York
ny_tz = tz.gettz("America/New_York")
# Crear una fecha y hora en Nueva York
fecha_ny = datetime(2023, 10, 1, 15, 23, 25, tzinfo=ny_tz)
```
