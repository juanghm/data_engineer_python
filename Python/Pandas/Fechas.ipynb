# Manipulando fechas con Pandas

## Timezones y Fechas
Se puede utilizar la función `pd.to_datetime()` para convertir cadenas de texto a objetos de fecha y hora. Pandas maneja automáticamente las zonas horarias, permitiendo convertir fechas entre diferentes zonas horarias.

```python
import pandas as pd

# Crear un DataFrame con fechas y zonas horarias
data = {
    'fecha': ['2023-01-01 12:00:00', '2023-01-02 15:30:00'],
    'zona_horaria': ['America/New_York', 'Europe/Madrid']
}

df = pd.DataFrame(data)

# Convertir la columna 'fecha' a tipo datetime y establecer la zona horaria
df['fecha'] = pd.to_datetime(df['fecha']).dt.tz_localize('UTC')

# Convertir a la zona horaria especificada
df['fecha_nueva'] = df.apply(lambda row: row['fecha'].tz_convert(row['zona_horaria']), axis=1)

print(df[['fecha', 'zona_horaria', 'fecha_nueva']])

# Obtener el año de cada fecha.
df['año'] = df['fecha_nueva'].dt.year

# Obtener el nombre del dia de cada fecha.
df['dia_nombre'] = df['fecha_nueva'].dt.day_name()
df['dia_nombre'] = df['fecha_nueva'].dt.weekday_name()
```

## Fechas Ambiguas por Timezones
Pandas maneja automáticamente las zonas horarias, pero en algunos casos puede haber ambigüedad, especialmente durante el horario de verano. Se puede utilizar el parámetro `ambiguous` para resolver estas ambigüedades.

```python
import pandas as pd

# Crear un DataFrame con fechas ambiguas
data = {
    'fecha': ['2023-03-12 01:30:00', '2023-11-05 01:30:00'],
    'zona_horaria': ['America/New_York', 'America/New_York']
}

df = pd.DataFrame(data)

# Convertir la columna 'fecha' a tipo datetime y establecer la zona horaria
df['fecha'] = pd.to_datetime(df['fecha']).dt.tz_localize('America/New_York', ambiguous='NaT') # Si la fecha es ambigua, se asigna NaT
```
