# Shell

## Listar archivos ls
El comando `ls` se utiliza para listar los archivos y directorios en el sistema de archivos. La sintaxis básica es:
```bashls [opciones] [ruta]
```
- `ruta`: es la ubicación del directorio que deseas listar. Si no se especifica, se listará el directorio actual.
- Algunas opciones comunes incluyen:
  - `-l`: muestra una lista detallada con permisos, propietario, tamaño y fecha de modificación.
  - `-a`: muestra todos los archivos, incluidos los ocultos (los que comienzan con un punto).
  - `-h`: muestra los tamaños de archivo en un formato legible para humanos (por ejemplo, KB, MB).
  - `-R`: lista los subdirectorios de forma recursiva.
  - `-F`: agrega un indicador al final de cada nombre de archivo para mostrar su tipo (por ejemplo, `/` para directorios, `*` para archivos ejecutables).

## Copiar archivos cp
Para copiar archivos en Linux se utiliza el comando `cp`. La sintaxis básica es:
```bash
cp [opciones] origen destino
```
- `origen`: es el archivo o directorio que deseas copiar.
- `destino`: es la ubicación donde deseas copiar el archivo o directorio.
- Para copiar varios archivos, puedes listar múltiples archivos de origen seguidos del destino.

## Mover archivos mv
El comando `mv` se utiliza para mover o renombrar archivos y directorios. La sintaxis básica es:
```bash
mv [opciones] origen destino
```
- `origen`: es el archivo o directorio que deseas mover o renombrar.
- `destino`: es la nueva ubicación o el nuevo nombre del archivo o directorio.

## Renombrar archivos mv
El comando `mv` también se utiliza para renombrar archivos. La sintaxis es la misma que para mover archivos:
```bash
mv [opciones] nombre_actual nuevo_nombre
```
- `nombre_actual`: es el nombre actual del archivo o directorio.
- `nuevo_nombre`: es el nuevo nombre que deseas asignar al archivo o directorio.

## Eliminar archivos rm
El comando `rm` se utiliza para eliminar archivos y directorios. La sintaxis básica es:
```bash
rm [opciones] archivo
```
- `archivo`: es el archivo o directorio que deseas eliminar.
- Para eliminar directorios y su contenido de forma recursiva, puedes usar la opción `-r`.

## Imprimir el contenido de un archivo cat
El comando `cat` se utiliza para mostrar el contenido de un archivo en la terminal. La sintaxis básica es:
```bash
cat archivo
```
- `archivo`: es el archivo cuyo contenido deseas ver.

## Ver el contenido de un archivo línea por línea less
El comando `less` se utiliza para ver el contenido de un archivo página por página. La sintaxis básica es:
```bash
less archivo
```
- `archivo`: es el archivo cuyo contenido deseas ver.
- Puedes navegar por el archivo usando las teclas de flecha, y salir presionando `q`.
- `less` es especialmente útil para archivos grandes, ya que no carga todo el archivo en memoria de una vez.
- A diferencia de `cat`, `less` permite desplazarse hacia adelante y hacia atrás en el archivo.
- También puedes buscar texto dentro del archivo presionando `/` seguido del término de búsqueda.
- Para avanzar una página, presiona la barra espaciadora, y para retroceder una página, presiona `b`.

## Mostrar las primeras líneas de un archivo head
El comando `head` se utiliza para mostrar las primeras líneas de un archivo. La sintaxis básica es:
```bash
head [opciones] archivo
```
- `archivo`: es el archivo cuyo contenido deseas ver.
- Por defecto, `head` muestra las primeras 10 líneas del archivo.
- Puedes especificar el número de líneas a mostrar usando la opción `-n`. Por ejemplo, `head -n 5 archivo` mostrará las primeras 5 líneas del archivo.
- `!head` es útil para obtener una vista rápida del contenido de un archivo sin abrirlo por completo.

## Mostrar las últimas líneas de un archivo tail
El comando `tail` se utiliza para mostrar las últimas líneas de un archivo. La sintaxis básica es:
```bash
tail [opciones] archivo
```
- `archivo`: es el archivo cuyo contenido deseas ver.

## Extraer secciones específicas de un archivo cut
El comando `cut` se utiliza para extraer secciones específicas de cada línea de un archivo. La sintaxis básica es:
```bash
cut [opciones] archivo
```
- `archivo`: es el archivo del cual deseas extraer secciones.
- Algunas opciones comunes incluyen:
  - `-f`: especifica los campos a extraer (por ejemplo, `-f 1,3` extrae los campos 1 y 3).
  - `-d`: especifica el delimitador que separa los campos (por defecto es la tabulación).
  - `-c`: especifica los caracteres a extraer (por ejemplo, `-c 1-5` extrae los primeros 5 caracteres de cada línea).

## Ver el historial de comandos history
El comando `history` se utiliza para mostrar el historial de comandos ejecutados en la terminal.
La sintaxis básica es:
```bash
history [opciones]
```
- No requiere argumentos, simplemente muestra una lista numerada de los comandos ejecutados recientemente.

## Buscar texto en archivos grep
El comando `grep` se utiliza para buscar texto dentro de archivos. La sintaxis básica es:
```bash
grep [opciones] patrón archivo
```
- `patrón`: es el texto o expresión regular que deseas buscar.
- `archivo`: es el archivo en el que deseas buscar.
- Algunas opciones comunes incluyen:
  - `-i`: ignora mayúsculas y minúsculas al buscar.
  - `-r`: busca de forma recursiva en directorios.
  - `-v`: muestra las líneas que no coinciden con el patrón.
  - `-n`: muestra el número de línea donde se encuentra el patrón.

## Combinar archivos paste
El comando `paste` se utiliza para combinar líneas de varios archivos lado a lado. La sintaxis básica es:
```bash
paste [opciones] archivo1 archivo2 ...
```

## Combinar varios comandos usando pipes
Los pipes (`|`) se utilizan para combinar la salida de un comando como entrada de otro.
Por ejemplo:
```bash
comando1 | comando2 | comando3
```
- Esto permite encadenar comandos para realizar operaciones más complejas.
- Por ejemplo, puedes usar `cut -d , -f 2 seasonal/summer.csv | grep -v Tooth"` para extraer el segundo campo de un archivo CSV y luego filtrar las líneas que no contienen la palabra "Tooth".

## Contar líneas wc
El comando `wc` se utiliza para contar líneas, palabras y caracteres en un archivo. La sintaxis básica es:
```bash
wc [opciones] archivo
```
- `archivo`: es el archivo cuyo contenido deseas contar.
- Algunas opciones comunes incluyen:
    - `-l`: cuenta solo las líneas.
    - `-w`: cuenta solo las palabras.
    - `-c`: cuenta solo los caracteres.

## Ordenar archivos sort
El comando `sort` se utiliza para ordenar líneas de texto en un archivo. La sintaxis básica es:
```bash
sort [opciones] archivo
```
- `archivo`: es el archivo cuyas líneas deseas ordenar.
- Algunas opciones comunes incluyen:
  - `-r`: ordena en orden inverso.
  - `-n`: ordena numéricamente.
  - `-k`: especifica la clave de ordenación (por ejemplo, `-k 2` ordena por la segunda columna).
  - `-u`: elimina líneas duplicadas.

## Wildcard
Los wildcards son caracteres especiales que se utilizan para representar uno o más caracteres en nombres de archivos y directorios. Los más comunes son:
- `*`: representa cero o más caracteres. Por ejemplo, `*.txt` coincide con todos los archivos que tienen la extensión `.txt`.
- `?`: representa un solo carácter. Por ejemplo, `file?.txt` coincide con `file1.txt`, `file2.txt`, etc.
- `[abc]`: coincide con cualquier carácter dentro de los corchetes. Por ejemplo, `file[12].txt` coincide con `file1.txt` y `file2.txt`.
- `{a,b}`: coincide con cualquiera de los elementos dentro de las llaves. Por ejemplo, `file{1,2}.txt` coincide con `file1.txt` y `file2.txt`.
- `!`: se utiliza para negar una coincidencia. Por ejemplo, `!(*.txt)` coincide con todos los archivos que no tienen la extensión `.txt`.
- `[]`: se utiliza para definir un rango de caracteres. Por ejemplo, `file[a-z].txt` coincide con `filea.txt`, `fileb.txt`, etc.
- `\`: se utiliza para escapar caracteres especiales. Por ejemplo, si deseas buscar un archivo que contenga un asterisco en su nombre, puedes usar `file\*.txt`.
- `~`: se utiliza para representar el directorio home del usuario actual. Por ejemplo, `~/documentos` se refiere al directorio `documentos` dentro del directorio home del usuario actual.
- `.`: se utiliza para representar el directorio actual. Por ejemplo, `./archivo.txt` se refiere al archivo `archivo.txt` en el directorio actual.

## Storing Information in Variables
Se puede almacenar información en variables para su uso posterior en la línea de comandos. La sintaxis básica es:
```bash
variable=valor
```
Se asigna el valor a la variable sin espacios alrededor del signo igual. Para acceder al valor de la variable, se utiliza el símbolo `$` antes del nombre de la variable. Por ejemplo:
```bash
echo $variable
```

## Loops de comandos
Los loops de comandos permiten ejecutar un conjunto de comandos varias veces. En Bash, se pueden utilizar `for`, `while` y `until` para crear loops.
```bash
# Loop for
for i in {1..5}; do
    echo "Número: $i"
done

# Loop while
contador=1
while [ $contador -le 5 ]; do
    echo "Contador: $contador"
    contador=$((contador + 1))
done

# Loop until
contador=1
until [ $contador -gt 5 ]; do
    echo "Contador: $contador"
    contador=$((contador + 1))
done
```

## Editar archivos con nano
El editor de texto `nano` se utiliza para editar archivos de texto en la terminal. La sintaxis básica es:
```bash
nano archivo
```
- `archivo`: es el archivo que deseas editar. Si el archivo no existe, `nano` lo creará.
- Algunos atajos útiles en `nano` incluyen:
  - `Ctrl + O`: guardar el archivo.
  - `Ctrl + X`: salir de `nano`.
  - `Ctrl + K`: cortar una línea.
  - `Ctrl + U`: pegar una línea cortada.
  - `Ctrl + W`: buscar texto en el archivo.
  - `Ctrl + G`: mostrar la ayuda de `nano`.
